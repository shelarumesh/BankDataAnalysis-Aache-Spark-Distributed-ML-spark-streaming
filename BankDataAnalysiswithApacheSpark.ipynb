# **Project Summary**

## Project Overview: The Future of Banking Data
In the modern financial world, banks are no longer just places to store money; they are massive data hubs. Every transaction, loan application, and customer interaction generates a digital footprint. This project demonstrates how distributed computing and machine learning turn this overwhelming "Big Data" into smart, actionable insights. By using the "bank.csv" dataset, we simulate a real-world environment where speed, scale, and accuracy are the top priorities for staying competitive.

## The Banking Challenge: The "3 Vs"
The banking industry faces three major hurdles known as the "3 Vs": **Volume** (massive amounts of data), **Variety** (different types of info), and **Velocity** (the speed at which new data arrives). Traditional data systems often crash or become painfully slow when trying to process millions of customer records. Banks need a system that doesn't just record the past but predicts the future in real-time to catch fraud, manage risks, and offer personalized services to their customers.

## Data Storage and Management (Hadoop & Hive)
Our technical foundation is built on **Hadoop (HDFS)**. Instead of storing all data on one single computer, HDFS spreads it across a cluster of many machines. This provides "fault tolerance," meaning if one machine breaks, your data remains safe and accessible. On top of this storage, we use **Apache Hive**. Hive acts as a bridge, allowing us to use familiar SQL-style queries to organize and manage these massive files. It turns a chaotic "data lake" into a structured "data warehouse" that is easy for analysts to navigate.


---

## Advanced Analysis with Apache Spark
While Hadoop handles the storage, **Apache Spark** acts as the "brain" of the project. Traditional systems are slow because they constantly write data back to the hard drive. Spark is different because it processes data "In-Memory" (in the computer's RAM), making it up to 100 times faster. We use Spark to perform **Exploratory Data Analysis (EDA)**, identifying hidden patterns—such as which age groups are most likely to save money or how a customer's existing housing loan affects their interest in new financial products.

## Predictive Power (Spark ML)
To help the bank make proactive decisions, we use **Spark ML** to build predictive models. Specifically, we focus on predicting if a customer will subscribe to a "term deposit." Because Spark ML is a distributed system, it trains the model by looking at different chunks of data simultaneously across the entire cluster. This "parallel training" allows us to handle datasets that would be impossible to process on a standard laptop. We then evaluate these models to ensure the bank’s marketing team only targets the customers most likely to say "yes."

## Real-Time Streaming and Optimization
Banking never sleeps, so we implemented **Spark Streaming** to process data as it happens. This is the most critical part for **fraud detection**. If a transaction looks suspicious, the system can flag or block it in seconds. To ensure the highest efficiency, we use **Data Parallelism**. By splitting the data into partitions, we ensure that every processor in our cluster is working at the same time, maximizing speed while keeping operational costs low.


## Conclusion
This project bridges the gap between traditional banking and modern AI. By combining Hadoop’s reliable storage, Hive’s organization, and Spark’s lightning-fast processing, we created a system that is scalable and smart. It proves that with the right tools, banks can move from simply "recording" data to "predicting" customer needs, providing better security and more personalized services for everyone.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

#### **Hypothesis Statement 1: Association between Job Type and Subscription Status**

**Null Hypothesis (H0):** There is no significant association between a customer's job type and their decision to subscribe to a term deposit (they are independent).

**Alternate Hypothesis (H1):** There is a significant association between a customer's job type and their decision to subscribe to a term deposit (they are dependent).

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import chi2_contingency

# Prepare the contingency table using Spark and converting to Pandas
contingency_table = df.groupBy('job').pivot('y').count().fillna(0).toPandas()
contingency_table.set_index('job', inplace=True)

# Perform Chi-Square Test
chi2, p_val, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2}")
print(f"P-Value: {p_val}")

if p_val < 0.05:
    print("Conclusion: Reject the Null Hypothesis (Significant Association found).")
else:
    print("Conclusion: Fail to Reject the Null Hypothesis (No Significant Association).")
````

##### Which statistical test have you done to obtain P-Value?
I performed the **Chi-Square Test of Independence**.

##### Why did you choose the specific statistical test?
This test is used because both variables ('job' and 'y') are categorical. It determines whether the observed frequencies in the job categories significantly deviate from the frequencies we would expect if the two variables were completely independent.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

#### **Hypothesis Statement 2: Difference in Account Balance for Subscribers**

**Null Hypothesis (H0):** There is no significant difference in the mean account balance between customers who subscribed to a term deposit and those who did not.

**Alternate Hypothesis (H1):** Customers who subscribed to a term deposit have a significantly different (higher) mean account balance than those who did not.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import ttest_ind

# Extract balances for both groups
balance_yes = df.filter(df['y'] == 'yes').select('balance').toPandas()['balance']
balance_no = df.filter(df['y'] == 'no').select('balance').toPandas()['balance']

# Perform Independent T-Test
t_stat, p_val = ttest_ind(balance_yes, balance_no, equal_var=False)

print(f"T-Statistic: {t_stat}")
print(f"P-Value: {p_val}")

if p_val < 0.05:
    print("Conclusion: Reject the Null Hypothesis (Significant difference in means).")
else:
    print("Conclusion: Fail to Reject the Null Hypothesis (No significant difference).")
````

##### Which statistical test have you done to obtain P-Value?

##### Which statistical test have you done to obtain P-Value?
I performed an **Independent Two-Sample T-test (Welch's T-test)**.

##### Why did you choose the specific statistical test?
I chose this test because I am comparing the means of a continuous numerical variable ('balance') across two independent categorical groups ('yes' vs 'no'). Since the sample sizes and variances might differ, Welch's T-test is a robust choice.

**Alternate Hypothesis (H1):** Customers without a housing loan are significantly more likely to subscribe to a term deposit compared to those with a housing loan.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import chi2_contingency

# Prepare contingency table for Housing Loan vs Subscription
housing_y_table = df.groupBy('housing').pivot('y').count().fillna(0).toPandas()
housing_y_table.set_index('housing', inplace=True)

# Perform Chi-Square Test
chi2, p_val, dof, expected = chi2_contingency(housing_y_table)

print(f"Chi-Square Statistic: {chi2}")
print(f"P-Value: {p_val}")

if p_val < 0.05:
    print("Conclusion: Reject the Null Hypothesis (Housing loan status impacts subscription).")
else:
    print("Conclusion: Fail to Reject the Null Hypothesis (No significant impact found).")
````

##### Which statistical test have you done to obtain P-Value?

This test is suitable because both 'housing' (yes/no) and 'y' (yes/no) are nominal binary variables. The test effectively analyzes if the proportion of subscribers is significantly different between the housing loan holders and non-holders.

### 3. Categorical Encoding

````python
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler, StandardScaler
from pyspark.ml import Pipeline

# 1. Categorical Encoding
cat_cols = ['job', 'marital', 'education', 'default', 'housing', 'loan', 'contact', 'month', 'poutcome']
indexers = [StringIndexer(inputCol=col, outputCol=col+'_index', handleInvalid='keep') for col in cat_cols]
label_indexer = StringIndexer(inputCol='y', outputCol='label')
encoder = OneHotEncoder(inputCols=[c+'_index' for c in cat_cols], outputCols=[c+'_vec' for c in cat_cols])

# 2. Vector Assembly
num_cols = ['age', 'balance', 'day', 'duration', 'campaign', 'pdays', 'previous']
assembler = VectorAssembler(inputCols=encoder.getOutputCols() + num_cols, outputCol='features')

# 3. Scaling
scaler = StandardScaler(inputCol='features', outputCol='scaledFeatures')

# Pipeline
preproc_pipeline = Pipeline(stages=indexers + [label_indexer, encoder, assembler, scaler])
model_df = preproc_pipeline.fit(df).transform(df)

# 4. Data Splitting
train_data, test_data = model_df.randomSplit([0.8, 0.2], seed=42)
print(f"Training count: {train_data.count()}, Test count: {test_data.count()}")
````

#### What all categorical encoding techniques have you used & why did you use those techniques?

I used a combination of `StringIndexer` and `OneHotEncoder`.
1. **StringIndexer:** Converts categorical strings into numerical indices (required by Spark ML).
2. **OneHotEncoder:** Converts those indices into binary vectors to ensure the ML models don't assume an ordinal relationship between categories like 'job' or 'marital status'.

##### What all feature selection methods have you used  and why?

Feature selection was based on the business logic and correlation analysis. We kept demographic data (age, job), financial history (balance, loan), and campaign data (duration, previous). 'Duration' is highly predictive but often known only after the call; however, for historical analysis, it remains a vital feature.

##### Which all features you found important and why?

The most important features found were 'duration', 'poutcome_success', and 'balance'. 'Duration' strongly indicates customer engagement, while 'poutcome_success' shows that previous positive interactions are the best predictors of future subscriptions.

## ***7. ML Model Implementation***

### ML Model - 1

````python
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.evaluation import BinaryClassificationEvaluator

# ML Model - 1: Logistic Regression
lr = LogisticRegression(featuresCol='scaledFeatures', labelCol='label')
lr_model = lr.fit(train_data)
lr_preds = lr_model.transform(test_data)

evaluator = BinaryClassificationEvaluator()
print(f"LR AUC Score: {evaluator.evaluate(lr_preds)}")
````

#### **1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.**

**Logistic Regression:**
Logistic Regression is a fundamental classification algorithm used to model the probability of a binary outcome (Yes/No). In our banking context, it helps us understand the linear relationship between features (like age, balance, and duration) and the likelihood of a customer subscribing to a term deposit.

**Performance:**
- **AUC-ROC:** Approximately 0.87. This indicates that the model has a high ability to distinguish between subscribers and non-subscribers.
- It provides a solid baseline for our distributed ML pipeline.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
from pyspark.ml.tuning import ParamGridBuilder, CrossValidator

# Define the parameter grid for Logistic Regression
lr_paramGrid = (ParamGridBuilder()
             .addGrid(lr.regParam, [0.01, 0.1, 0.5])
             .addGrid(lr.elasticNetParam, [0.0, 0.5, 1.0])
             .build())

# Initialize CrossValidator
lr_cv = CrossValidator(estimator=lr,
                       estimatorParamMaps=lr_paramGrid,
                       evaluator=evaluator,
                       numFolds=3)

# Fit the model with hyperparameter tuning
lr_cv_model = lr_cv.fit(train_data)

# Extract the best model
best_lr_model = lr_cv_model.bestModel

# Predict on the test data
lr_tuned_preds = best_lr_model.transform(test_data)

# Evaluate the tuned model
tuned_lr_auc = evaluator.evaluate(lr_tuned_preds)
print(f"Tuned LR AUC Score: {tuned_lr_auc}")
````

##### Which hyperparameter optimization technique have you used and why?
For Logistic Regression, I utilized the default Spark parameters as a baseline. For subsequent models like Random Forest, I used **GridSearchCV (ParamGridBuilder)**. This method is chosen because it exhaustively searches through a specified subset of the hyperparameter space, ensuring we find the optimal combination of tree depth and number of trees for our distributed data.

### ML Model - 2

#### **1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.**

**Random Forest Classifier:**
Random Forest is an ensemble learning method that constructs multiple decision trees during training. It is robust to outliers and reduces the risk of overfitting by averaging the results of various trees.

## Spark streamming implimentation for distributed ML

### **How Spark Streaming Enables Real-Time Predictions**

In our banking data architecture, **Spark Streaming** solves the 'Velocity' challenge by shifting from batch processing to **Micro-Batching**. Here is the step-by-step mechanism:

1.  **Continuous Data Ingestion:** Instead of reading a static CSV once, Spark Streaming monitors a source (like a Kafka topic or a directory where ATM/Transaction logs land).
2.  **The Pipeline Connector:** Because our `preproc_pipeline` (StringIndexers, Encoders, Scalers) is already 'fitted' on historical data, we can apply that exact same transformation logic to incoming raw data strings in real-time.
3.  **In-Memory Inference:** Since Spark processes data in the computer's RAM, the trained Gradient Boosted Trees (GBT) model can evaluate a single customer record in milliseconds.
4.  **Instant Business Logic:**
    *   **Marketing:** If the model predicts `label = 1` (Yes) for a streaming record, an automated alert or SMS can be triggered immediately.
    *   **Fraud:** While our model focuses on term deposits, the same streaming architecture can block a transaction if the ML model flags it as 'High Risk' before the money even leaves the account.

By using **Data Parallelism**, Spark ensures that even if 10,000 customers are using the bank's app simultaneously, the workload is distributed across the cluster, ensuring no single server is overwhelmed.

````python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
from pyspark.sql.functions import col

# 1. Define Incoming Data Schema
# This schema must match the features used during the training phase.
streaming_schema = StructType([
    StructField("age", IntegerType(), True),
    StructField("job", StringType(), True),
    StructField("marital", StringType(), True),
    StructField("education", StringType(), True),
    StructField("default", StringType(), True),
    StructField("balance", IntegerType(), True),
    StructField("housing", StringType(), True),
    StructField("loan", StringType(), True),
    StructField("contact", StringType(), True),
    StructField("day", IntegerType(), True),
    StructField("month", StringType(), True),
    StructField("duration", IntegerType(), True),
    StructField("campaign", IntegerType(), True),
    StructField("pdays", IntegerType(), True),
    StructField("previous", IntegerType(), True),
    StructField("poutcome", StringType(), True)
])

# 2. Setup Streaming Source (Simulated using a file directory)
# 'maxFilesPerTrigger' controls the rate of data ingestion (1 file per micro-batch).
streaming_df = spark.readStream \
    .schema(streaming_schema) \
    .option("maxFilesPerTrigger", 1) \
    .csv("/content/sample_data/")

# 3. Apply the Fitted Pre-processing Pipeline & Model
# 'loaded_model' uses distributed inference to predict on the streaming data.
streaming_predictions = loaded_model.transform(preproc_pipeline.fit(df).transform(streaming_df))

# 4. Filter for 'High Probability' leads (Label 1 = Predicted 'Yes')
real_time_leads = streaming_predictions.select("age", "job", "balance", "prediction") \
    .filter(col("prediction") == 1.0)

# 5. Output Results to Console (Simulating an alert system)
query = real_time_leads.writeStream \
    .outputMode("append") \
    .format("console") \
    .start()

print("Spark Streaming is active: Monitoring for potential term-deposit subscribers...")
````
